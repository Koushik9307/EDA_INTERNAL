check the unique values in the plan type and replace the null values with the mode

```python
df["plan_type"].unique()
```

```python
df["plan_type"] = df["plan_type"].fillna(df["plan_type"].mode()[0])
df["plan_type"] = df["plan_type"].str.lower().str.strip()

df["plan_type"] = df["plan_type"].replace({

    'basic': 'Basic',
    'bsic': 'Basic',
    'basi': 'Basic',
    'basc': 'Basic',
    'baic': 'Basic',
    'asic': 'Basic',

    'standard': 'Standard',
    'standrd': 'Standard',
    'standar': 'Standard',
    'stndard': 'Standard',
    'sandard': 'Standard',
    'tandard': 'Standard',
    'stadard': 'Standard',

    'unlimited': 'Unlimited',
    'unliited': 'Unlimited',
    'nlimited': 'Unlimited',
    'ulimited': 'Unlimited',
    'unimited': 'Unlimited',
    'unlimitd': 'Unlimited',
    'unlmited': 'Unlimited'
})
```

check the unique values in the has family plan and replace the nan values with the mode and handle the inconsistancy data

```python
df["has_family_plan"].unique()
```

```python
df["has_family_plan"] = df["has_family_plan"].str.lower().str.strip()

df["has_family_plan"] = df["has_family_plan"].replace({

    'yes': 'Yes',
    'ys': 'Yes',
    'ye': 'Yes',
    'es': 'Yes',

    'no': 'No',
    'n': 'No',
    'o': 'No'
})
df["has_family_plan"] = df["has_family_plan"].fillna(df["has_family_plan"].mode()[0])
```

to check weather the data contains the outliers or not

```python
import matplotlib.pyplot as plt
import seaborn as sns
num_cols = [
    "monthly_minutes",
    "monthly_data_mb",
    "monthly_sms",
    "contract_months",
    "overdue_payments",
]

plt.figure(figsize=(15, 8))
for i, col in enumerate(num_cols, 1):
  plt.subplot(2, 3, i)
  sns.boxplot(y=df[col])
  plt.title(f"Boxplot of {col}")

plt.tight_layout()
plt.show()
```

removing outlier rows from the data set

```python

outlier_cols = ["monthly_minutes", "monthly_data_mb"]

for col in outlier_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1

  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]


print(df.shape)
```

UNIVARIATE ANALYSIS:barchart ,pie chart and histogrm on the above data columns

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6, 4))
sns.histplot(df["monthly_minutes"], kde=True, bins=20, color="skyblue")
plt.title("Histogram of Monthly Minutes")
plt.show()

plt.figure(figsize=(6, 4))
sns.countplot(
    x=df["plan_type"], hue=df["plan_type"], palette="Set2", legend=False
)
plt.title("Bar Chart of Plan Type")
plt.show()

plt.figure(figsize=(5, 5))
counts = df["has_family_plan"].value_counts()
plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2"),
)
plt.title("Pie Chart of Family Plan")
plt.show()
```

BIVARIATE ANALYSIS: boxplot,grouped bar chart and bubble chart on the columns

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6, 4))
sns.scatterplot(
    data=df,
    x="monthly_minutes",
    y="monthly_data_mb",
    hue="churn_next_month",
    palette="viridis",
)
plt.title("Monthly Minutes vs Monthly Data (Colored by Churn)")
plt.show()

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="plan_type", hue="has_family_plan", palette="Set2")
plt.title("Plan Type by Family Plan Status")
plt.show()


plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x="plan_type", y="monthly_minutes", palette="Set2")
plt.title("Monthly Minutes Distribution across Plan Types")
plt.show()
```
